```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.font_manager import FontProperties
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.backends.backend_pdf import PdfPages
```

```python
simulation = 'cells'
word_path = 'E'
dir_dataset = word_path + ':/DataSet5GHz/'
n_APs = [1,2,3,4,5]
K = 1.380649e-23
T = 290
B = 80
v_min = K*T*B
v_max = 26
```

```python
mean_loss = []
mean_val_loss = []
   
mean_rmse = []
mean_val_rmse = []
mean_mae = []
mean_val_mae = []
   
mean_error = []
mean_val_error = []
mean_iou = []
mean_val_iou = []
```

```python
for i in n_APs:
    
    if i != 1 and simulation == 'cells' or simulation == 'maps':
        loaded_arrays = np.load(dir_dataset + '/Models/' + str(i) + 'AP/' + simulation + '_' + str(i) +'_history.npz')
        
        losses = loaded_arrays['losses']
        val_losses = loaded_arrays['val_losses']

        if simulation == 'maps':
            rmse_scores = loaded_arrays['rmse_scores']
            val_rmse_scores = loaded_arrays['val_rmse_scores']
            mae_scores = loaded_arrays['mae_scores']
            val_mae_scores = loaded_arrays['val_mae_scores']
            
        if simulation == 'cells':
            error_scores = loaded_arrays['pixel_e_scores']
            val_error_scores = loaded_arrays['val_pixel_e_scores']
            iou_scores = loaded_arrays['iou_scores']
            val_iou_scores = loaded_arrays['val_iou_scores']
            
        mean_loss.append(np.mean(losses, axis=0))
        mean_val_loss.append(np.mean(val_losses, axis=0))

        if simulation == 'maps':    
            mean_rmse.append(np.mean(rmse_scores, axis=0))
            mean_val_rmse.append(np.mean(val_rmse_scores, axis=0))
            mean_mae.append(np.mean(mae_scores, axis=0))
            mean_val_mae.append(np.mean(val_mae_scores, axis=0))
            
        if simulation == 'cells':    
            mean_error.append(np.mean(error_scores, axis=0))
            mean_val_error.append(np.mean(val_error_scores, axis=0))
            mean_iou.append(np.mean(iou_scores, axis=0))
            mean_val_iou.append(np.mean(val_iou_scores, axis=0))
```

```python
def create_subplots(mean_m1, mean_val_m1, mean_m2, mean_val_m2, simulation = simulation):
    if simulation == 'maps':
        fig, axs = plt.subplots(3, 2, figsize=(9,9))
    else:
        fig, axs = plt.subplots(2, 2, figsize=(9, 9))
        
    colors = ['blue', 'blue', 'red', 'red']
    linestyles = ['-', '--', '-', '--']
    data = [mean_m1, mean_val_m1, mean_m2, mean_val_m2]
    
    if simulation == 'maps':
        
        labels = ['RMSE', 'RMSE_val', 'MAE', 'MAE_val']        
    else:
        labels = ['PER', 'PER_val', 'IoU', 'IoU_val'] 
        

    # for i, ax in enumerate(axs.flat[:-1]):
    for i, ax in enumerate(axs.flat):
        if simulation == 'maps':
            for j in range(0, len(data), 2):
                ax.plot(data[j][i, :]*120.94, label=labels[j], color=colors[j], linestyle=linestyles[j])
                ax.plot(data[j+1][i, :]*120.94, label=labels[j+1], color=colors[j+1], linestyle=linestyles[j+1])
            ax.set_title(f'{i+1} APs')
        else:
            for j in range(0, len(data), 2):
                ax.plot(data[j][i, :]*100, label=labels[j], color=colors[j], linestyle=linestyles[j])
                ax.plot(data[j+1][i, :]*100, label=labels[j+1], color=colors[j+1], linestyle=linestyles[j+1])
            ax.set_title(f'{i+2} APs')
        # ax.set_ylabel('dBm')
        ax.grid(True)
        ax.legend(prop={'size': 14})
        
        if simulation =='maps':
            # plt.xlabel('Epochs')
            # plt.ylabel('dBm')
            ax.set_xlabel('Epochs')
            ax.set_ylabel('dBm')
        else:
            ax.set_xlabel('Epochs')
            ax.set_ylabel('%')
    
    fig.tight_layout()

    if simulation == 'maps':
        fig.delaxes(axs[2, 1])
        fig.patch.set_visible(False)

    plt.rc('font', family='Times New Roman', size=15)

    plt.tight_layout()
    # fig.tight_layout(pad=2.0)
    if simulation == 'maps':
        with PdfPages('E:/DataSet5GHz/Models/RME_metrics.pdf') as pdf:
            pdf.savefig(fig, bbox_inches='tight')
    else:
        with PdfPages('E:/DataSet5GHz/Models/CME_metrics.pdf') as pdf:
            pdf.savefig(fig, bbox_inches='tight')
            
    plt.show()
```
